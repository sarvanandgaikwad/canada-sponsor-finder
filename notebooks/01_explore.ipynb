```python
import pandas as pd
from pathlib import Path

RAW = Path("../data/raw/")
files = sorted(RAW.glob("*.xlsx"))
for i, f in enumerate(files):
    print(i, f.name)
```

### Finding header row

```python
f = files[-1]
raw = pd.read_excel(f, header=None, nrows=15)
raw
```

```python
HEADER_ROW = 1
df = pd.read_excel(f, header=HEADER_ROW)
print(df.shape)
df.columns.tolist()
```

### Removing notes rows

```python
pd.set_option("display.max_colwidth", None)   # stop pandas cutting long text
notes = df[df["Program Stream"].isna()]
for text in notes["Province/Territory"]:
    print(text)
    print()
```

```python
data = df[df["Program Stream"].notna()].copy()
print(data.shape)
data.isna().sum()
```

```python
for p in data["Province/Territory"].unique():
    print(repr(p))
```

```python
print(data["Program Stream"].value_counts())
print()
print(data["Incorporate Status"].value_counts())
```

```python
tech_codes = ("21211", "21220", "21221", "21222", "21223", "21230", "21231", "21232", "21233", "21234", "21311")
tech = data[data["Occupation"].astype(str).str.startswith(tech_codes)]
print(len(tech), "tech rows")
tech["Program Stream"].value_counts()
```

```python
tech.groupby("Employer")["Approved Positions"].sum().sort_values(ascending=False).head(20)
```

```python
for col in data.select_dtypes(include="str").columns:
    data[col] = data[col].str.strip()
data["Approved LMIAs"] = data["Approved LMIAs"].astype(int)
data["Approved Positions"] = data["Approved Positions"].astype(int)
data["Province/Territory"].unique()
```

### Tech jobs by province

```python
tech = data[data["Occupation"].str.startswith(tech_codes)]
tech.groupby("Province/Territory")["Approved Positions"].sum().sort_values(ascending=False)
```

```python
atlantic = ["Nova Scotia", "New Brunswick", "Prince Edward Island", "Newfoundland and Labrador"]
tech[tech["Province/Territory"].isin(atlantic)][["Province/Territory", "Employer", "Address", "Occupation", "Program Stream", "Approved Positions"]]
```

```python
gts = tech[(tech["Program Stream"] == "Global Talent Stream") & (tech["Province/Territory"] != "Quebec")]
gts.groupby("Employer")["Approved Positions"].sum().sort_values(ascending=False).head(25)
```

### Row counter per file

```python
for f in files:
    df_tmp = pd.read_excel(f, header=1)
    df_tmp.columns = df_tmp.columns.str.strip()          # 2022 file has trailing spaces in column names
    data_tmp = df_tmp[df_tmp["Program Stream"].notna()]  # drop the notes rows at the bottom
    notes_count = len(df_tmp) - len(data_tmp)
    print(f"{f.name:35} total: {len(df_tmp):6,}   notes: {notes_count:3}   data rows: {len(data_tmp):6,}")
```
